```python
#ESERCIZIO 4: L'azienda Object SpA ha creato una lista di quanti oggetti ha venduto ogni mese nell'ultimo anno:
import numpy as np
lst = [2000, 5500, 7200, 4320, 1280, 1900, 2500, 3900, 6410, 8150, 7100, 5350]

#trasformiamola in un array NumPy (casting):
lst = np.array(lst)

#e rispondiamo alle domande del CEO della Object SpA:
#qual è stata la vendita massima mensile? E quella minima?
print('la vendita massima mensile è stata:', lst.max())
print('la vendita minima mensile è stata:', lst.min())

#quali sono le vendite mensili maggiori di 5000 oggetti? E quante ne sono?
lst_magg_5000 = lst[lst > 5000]
print('le vendite mensili maggiori di 5000 sono state:', lst_magg_5000)

lst_count_magg_5000 = len(lst_magg_5000)
print('il numero dei prodotti con più di 5000 vendite è:', lst_count_magg_5000)

#quali sono le vendite minori di 3000 oggetti?
lst_min_3000 = lst[lst < 3000]
print('le vendite mensili minori di 3000 sono state:', lst_min_3000)

#in media quanti oggetti sono stati venduti al mese?
media = lst.mean()
print('la media dei prodotti venduti è:', round(media))
```

```python
#ESERCIZIO 8: Scrivere un programma che
#in input acquisisce una lista di numeri e un numero K
#in output, dovrà restituire la media di tutti i numeri nella lista maggiori o uguali a K
#se non ce ne dovesse essere nessuno, dovrà stampare a schermo un messaggio adeguato.
testo = input("Inserisci una lista di numeri separata da spazio: ")
testo_numeri = testo.split()
numeri = []
for testo_numero in testo_numeri:
    numeri.append(int(testo_numero))

k = int(input("Inserisci il valore di k: "))
numeri_selezionati = []
for n in numeri:
    if n >= k:
        numeri_selezionati.append(n)

if len(numeri_selezionati) == 0:
    print("Non ci sono numeri maggiori o uguali a k.")
else:
    media = sum(numeri_selezionati) / len(numeri_selezionati)
    print("La media dei numeri >= k è: ", media)
```

```python
#Esercizio 12: Abbiamo una lista di stringhe di prezzi in dollari, che erroneamente sono stati scritti con il simbolo dell'euro:
prezzi = ["100 €", "200 €", "500 €", "10 €", "50 €", "70 €"]
#cambiare il simbolo dell'euro (€) in quello del dollaro ($) per ogni stringa nella lista; il risultato sarà memorizzato in un'altra lista.abs

prezzi_dollari = []
for prezzo in prezzi:
    prezzo_corretto = prezzo.replace("€", "$")
    prezzi_dollari.append(prezzo_corretto)

print(prezzi_dollari)
```

```python
#Esercizio 13: Abbiamo una lista di studenti:
studenti = ["Alex", "Bob", "Cindy", "Dan", "Emma", "Faith", "Grace", "Henry", "Isabelle", "John"]
#vogliamo dividere gli studenti in due squadre per un campionato di Uno nel seguente modo:
#selezioneremo i nomi in posizione pari per un squadra, e i nomi in posizione dispari per l'altra. Creiamo due liste per ogni squadra, e alla fine visualizziamole.
squadra_pari = []
squadra_dispari = []
for i in range(len(studenti)):
    if i % 2 == 0:
        squadra_pari.append(studenti[i])
    else:
        squadra_dispari.append(studenti[i])

print("Squadra pari: ", squadra_pari)
print("Squadra dispari: ", squadra_dispari)
```

```python
#Esercizio 14: Abbiamo una lista con i guadagni degli ultimi 12 mesi (supponiamo da Gennaio a Dicembre):
guadagni = [100, 90, 70, 40, 50, 80, 90, 120, 80, 20, 50, 50]
#dobbiamo confrontare, stampando tutto a video, il guadagno di ogni mese con la media dei guadagni precedenti, e specificare nell'output se il guadagno attuale è maggiore o minore della media dei precedenti.
somma_precedenti = 0
conteggio_precedenti = 0
for i in range(len(guadagni)):
    guadagno_attuale = guadagni[i]
    mese = i + 1
    if conteggio_precedenti == 0:
        print("Mese " + str(mese) + ": " + str(guadagno_attuale) + " € (nessuna media precedente)")
    else:
        media_prec = somma_precedenti / conteggio_precedenti
        if guadagno_attuale > media_prec:
            confronto = "maggiore"
        elif guadagno_attuale < media_prec:
            confronto = "minore"
        else:
            confronto = "uguale"
        print("Mese " + str(mese) + ": " + str(guadagno_attuale) +
              " € (media prec: " + str(round(media_prec, 2)) +
              " € - il guadagno attuale è " + confronto + ")")

    somma_precedenti += guadagno_attuale
    conteggio_precedenti += 1  
```
